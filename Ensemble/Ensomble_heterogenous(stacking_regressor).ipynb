# Stacking (regressor)

```python
from sklearn.ensemble import StackingRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.metrics import r2_score
```

```python
X, y = make_regression(
    n_samples = 500,
    n_features = 20,
    n_informative = 5,
    random_state = 42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.3, random_state = 42
)
```

```python
lr = LinearRegression()
dtr = DecisionTreeRegressor()
svr = SVR()
```

```python
stacking_reg = StackingRegressor(
    estimators = [
        ("lr", lr),
        ("dtr", dtr),
        ("svr", svr)
    ],
    cv = 5
)
```

```python
stacking_reg.fit(X_train, y_train)
```

```python
y_pred = stacking_reg.predict(X_test)
y_train_pred = stacking_reg.predict(X_train)      # checking this buz o/p is like overfitting
```

```python
print("r^2 score test: ", r2_score(y_test, y_pred))
print("r^2 score train: ", r2_score(y_train, y_train_pred))
```

# It is not overfitting case
