```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report, confusion_matrix
```

```python
iris = pd.read_csv("IRIS.csv")
iris.sample()
```

```python
# convert 3 species into 0, 1, 2
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
iris["species"] = le.fit_transform(iris["species"])
iris.sample(5)
```

```python
X = iris.drop("species", axis = 1)
y = iris["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.5, random_state = 42
)
```

```python
# Logistic regression
lr = LogisticRegression(max_iter = 1000)
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
```

```python
# KNN
knn = KNeighborsClassifier(n_neighbors = 5)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
```

```python
# Naive bayes (Gaussian)
gb = GaussianNB()
gb.fit(X_train, y_train)
y_pred = gb.predict(X_test)
```

```python
models = {"Logistic Regression": lr, "KNeighborsClassifier": knn, "GaussianNB": gb}

for name, model in models.items():
    y_pred = model.predict(X)
    print(f"\n {name}")
    print(f"accuracy score: {accuracy_score(y, y_pred)}")
    print(f"cm: {confusion_matrix(y, y_pred)}")
    print(f"classification report:\n {classification_report(y, y_pred)}")
```

```python
# Conclusion
# Logistic Regression(0.967), KNN(0.954), Naive(0.954)
# so the best model is Logistic Regression
```
