```python
from sklearn.svm import SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn import datasets
```

```python
df = datasets.load_diabetes(as_frame = True).frame
```

```python
df.head()
```

```python
X = df.drop("target", axis = 1)
y = df["target"]
```

```python
X.head()
```

```python
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.33, random_state = 42
)
```

```python
y_scalar = StandardScaler()

y_train_scaled = y_scalar.fit_transform(y_train.values.reshape(-1, 1)).ravel()
y_test_scaled = y_scalar.transform(y_test.values.reshape(-1, 1)).ravel()
```

```python
model = SVR()

model.fit(X_train, y_train_scaled)
```

```python
y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("test r^2 score", r2_score(y_test_scaled, y_test_pred_scaled))
print("train r^2 score", r2_score(y_train_scaled, y_train_pred_scaled))
```

```python
# Linear

model = SVR(kernel = "linear")
model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("test r^2 score", r2_score(y_test_scaled, y_test_pred_scaled))
print("train r^2 score", r2_score(y_train_scaled, y_train_pred_scaled))
```

```python
# polynomial
# defenatially a overfitting case buz value difference is too much between train and test r^2
model = SVR(kernel = "poly")
model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("test r^2 score", r2_score(y_test_scaled, y_test_pred_scaled))
print("train r^2 score", r2_score(y_train_scaled, y_train_pred_scaled))
```

# Hyperparameter tunning using gridsearchCV

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "C": [1, 2, 5, 10, 20, 50, 100],
    "kernel": ["rbf", "linear"],
    "epsilon": [0.01, 0.1, 0.2, 0.3, 0.5]    
}
```

```python
svr = SVR()

grid_search = GridSearchCV(svr, param_grid, scoring = "r2", cv = 5)
grid_search.fit(X_train, y_train_scaled)
```

```python
print("best params - ", grid_search.best_params_)
```

```python
best_model = SVR(
    kernel = "linear",
    C = 10,
    epsilon = 0.01
)

best_model.fit(X_train, y_train_scaled)

y_test_pred_scaled = best_model.predict(X_test)
y_train_pred_scaled = best_model.predict(X_train)

print("test r^2 score", r2_score(y_test_scaled, y_test_pred_scaled))
print("train r^2 score", r2_score(y_train_scaled, y_train_pred_scaled))
```

```python
# Linear SVR

from sklearn.svm import LinearSVR

model = LinearSVR(C = 10, epsilon = 0.1, max_iter = 2000)

model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("test r^2 score", r2_score(y_test_scaled, y_test_pred_scaled))
print("train r^2 score", r2_score(y_train_scaled, y_train_pred_scaled))
```
